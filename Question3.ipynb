```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread('img.jpg', cv.IMREAD_GRAYSCALE)

if img is not None:
    edges = cv.Canny(img, 550, 690)
    indices = np.where(edges != 0)
    x = indices[1]
    y = indices[0]

    m, b = np.polyfit(x, y, 1)

    x_fit = np.array([np.min(x), np.max(x)])
    y_fit = m * x_fit + b

    plt.figure(figsize=(6, 6))
    plt.scatter(x, y, s=1, c='blue', marker='.', label='Edge Points')
    plt.plot(x_fit, y_fit, color='red', linewidth=2, label=f'LS Fit: y={m:.2f}x+{b:.2f}')

    plt.gca().invert_yaxis()
    plt.title('Least-Squares-Fit Line on Extracted Edges')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()

    print(f"Estimated Slope (m): {m}")
    print(f"Estimated Intercept (b): {b}")
else:
    print("Error: Image not found.")
```
