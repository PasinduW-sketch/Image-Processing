Least Squares Fit

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
uploaded = files.upload()

filename = list(uploaded.keys())[0]
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

edges = cv2.Canny(img, 550, 690)

indices = np.where(edges != 0)
x = indices[1]   # horizontal coordinates
y = indices[0]   # vertical coordinates

coeffs = np.polyfit(x, y, 1)
m_ls = coeffs[0]
b_ls = coeffs[1]

theta_ls = np.degrees(np.arctan(m_ls))
print("Least Squares Angle (degrees):", theta_ls)

plt.figure(figsize=(8,5))
plt.scatter(x, y, s=1)
plt.plot(x, m_ls*x + b_ls, color='red', linewidth=2)
plt.gca().invert_yaxis()
plt.title(f"Least Squares Fit (θ ≈ {theta_ls:.2f}°)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
